```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('Echobox Assessment - Part 2 Dataset - Data (April 2021).csv')
```

```python
position_data = df['Title Type'].value_counts()
```

```python
position_data.plot(kind='pie',title='Position Types',autopct='%1.1f%%',legend=False)
```

```python
reply_data = df['Reply type'].value_counts()
```

```python
reply_data.plot(kind='pie',autopct='%1.1f%%',title='Reply Satstifaction',legend=True)
```

```python
region_data = df['Geographical Region'].value_counts()
```

```python
region_data.plot(kind='barh',color='blue',title='Marketing Regions',ylabel='Region',xlabel='Frequency')
```

The most popular region was North America because it is the biggest regions of the marketing industry.

```python
campaign_quarter_data = df['Campaign Name'].value_counts()
```

```python
campaign_quarter_data.plot(kind='barh',color='blue',xlabel='Frequency',ylabel='Campaign Name',title='Campaign Name Quarters')
```

This bar graph shows the most popular campaign name was 'Sales Emails - Q4 Jolene', the second biggest campaign is 'Sales Emails - Q4 Alexander'. Q4 Jolene has most number of roles, regions and replies. 'Sent Emails - Q4 Jacob has the smallest number than others.
