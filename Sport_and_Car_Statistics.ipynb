```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
```

```python
products = pd.read_excel('H+ Sport products.xlsx',sheet_name='Sheet1')
```

```python
products.info()
```

```python
products.head(10)
```

```python
sns.violinplot(data=products,x='Product category',y='Price',palette='Set1')
plt.title('Price distribution by Product Category')
plt.xlabel('Category')
plt.show()
```

In the violin graph, the most popular catergory was 'Fitness' because this category has the biggest number of products than other categories.

```python
employees = pd.read_excel('H+ Sport Employees.xlsx',sheet_name='Employees')
```

```python
employees.head(10)
```

```python
sns.violinplot(data=employees,x='Status',y='Job Rating',palette='rainbow')
plt.title('Job Satisfaction by Status')
plt.ylabel('Rating Scale')
plt.show()
```

```python
sns.violinplot(data=employees,x='Department',y='Job Rating',hue='Department')
plt.xticks(rotation=85)
plt.title('Job Satisfaction by Department')
plt.ylabel('Rating Scale')
plt.show()
```

```python
sns.violinplot(data=employees,x='Building',y='Job Rating',palette='Set1')
plt.title('Job Satisfaction by Building')
plt.ylabel('Rating Scale')
plt.show()
```

```python
sns.violinplot(data=employees,x='Status',y='Years',palette='Set1')
plt.title('Job Status')
plt.ylabel('Number of Years')
plt.show()
```

```python
sns.violinplot(data=employees,x='Department',y='Years',palette='Set1')
plt.title('Job Departments')
plt.ylabel('Number of Years')
plt.xticks(rotation=85)
plt.show()
```

```python
sns.violinplot(data=employees,x='Building',y='Years',palette='Set1')
plt.title('Job Buildings')
plt.ylabel('Number of Years')
plt.show()
```

```python
cars = pd.read_csv('car_financing.csv')
```

```python
cars.head(10)
```

```python
sns.violinplot(data=cars,x='car_type',y='Interest Paid',palette='Set1')
plt.xlabel('Car Type')
plt.title('Car Interest Paid')
```

In the violin graph, the most popular interest was 'Toyota Sienna' because this car type has largest number of interest paid.
The least popular interest was 'Toyota Carolla'.
Both makes of these cars are Toyota and Volkswagen.
Three Toyotas has 3 models and Volkswagen has only model called 'Golf R'

```python
sns.violinplot(data=cars,x='car_type',y='Principal Paid',palette='Set1')
plt.xlabel('Car Type')
plt.title('Car Principal Paid')
```

Most Popular Principal Paid was 'VW Golf R' than other car types.
