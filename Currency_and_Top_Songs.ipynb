```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('banknote-dimensions.csv')
```

```python
df.eval('area = length * width')
```

```python
currencyOne = df.groupby(['currency','code'])
```

```python
currencyOne.first()
```

```python
currencyTwo = df.groupby(['currency']).agg(total_value=('value','sum'),
                                                avg_length=('length','mean'),
                                                avg_width=('width','mean'))
```

```python
currencyTwo
```

```python
currencyThree = df.groupby(['currency']).agg(avg_value=('value','mean'))
```
