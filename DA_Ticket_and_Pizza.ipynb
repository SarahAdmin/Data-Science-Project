```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('ticket_data.csv',encoding='latin-1')
```

```python
ticket_count = df['Ticket price'].value_counts()
```

```python
ticket_count
```

```python
ticket_count.plot(kind='pie',legend=True,title='Ticket Prices',autopct='%1.1f%%')
```

The Pie chart shows the number of attendees purchased the ticket prices are £0 and £20. 82 attendees paid tickets for £20 and 21 attendees paid tickets for free of charge.

```python
df1 = pd.read_json('StudentPizzaOrder.json')
```

```python
final_toppings = df1['Topping'].value_counts()
```

```python
final_toppings.plot(kind='pie',autopct='%1.1f%%',legend=False,title='Pizza Topping Orders')
```

```python
final_toppings.plot(kind='barh',xlabel='Frequency',title='Pizza Orders')
```

On these charts (Pie and bar) are showing the pizza order data. The most popular topping is Meat Feast because students ordered Meat Feast pizza. The least popular toppings are BBQ Chicken and Cheese & Tomato but there are the same.
