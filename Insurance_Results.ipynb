```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
```

```python
members= pd.read_excel('member_benefits.xlsx',sheet_name='Data 1')
```

```python
members.info()
```

```python
members.duplicated().value_counts()
```

```python
members.columns
```

```python
sns.violinplot(data=members,x='Status',y='Age',palette='Set2')
plt.title('Status Distrbution per Age',fontsize=14,fontweight='bold')
plt.ylabel('Age')
plt.xlabel('Status')
plt.show()
```

```python
sns.histplot(data=members,x='Age',hue='Status',multiple='stack')
plt.title('Age Distribution per Status',fontsize=14,fontweight='bold')
```

```python
fig = plt.subplots(figsize=(6,10))
ax = sns.countplot(data=members,x='Status',palette='mako')
for container in ax.containers:
  ax.bar_label(container)
plt.title('Status Distribution',fontsize=14,fontweight='bold')
plt.show()
```

On the bar graph, the most popular status was 'In Payment' because 19,465 members currently in the 'In Payment' Status. 532 members currently in 'Active' status. Three members in 'Addition' status.
