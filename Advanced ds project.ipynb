```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
```

```python
feedback = pd.read_csv('October Feedback.csv')
```

```python
feedback.dtypes
```

```python
feedback.columns
```

```python
feedback.head(10)
```

```python
sns.histplot(data=feedback,x='How satisfied are you with the way you were treated? - 05c18285-2750-479f-8afc-d560b66f552c',hue='Location')
```

```python
app1 = sns.boxplot(data=feedback,x='How satisfied are you with the ease of the application process? - 513b2b55-b818-4f03-bf6f-e82b180c9e60',y='Team',palette='rainbow',hue='Location')
plt.xticks(rotation=90)
```

```python
Dept1 = sns.boxplot(data=feedback,x='Overall, how satisfied were you with your interviewing experience? - 693b0a81-d4e7-40a2-9135-7fc38fa43eaf',hue='Department',linewidth=0.5)
Dept1.set(title='October Feedback 1',xlabel='How satisfied were you with you interviewing experience?')
plt.xticks(rotation=90)
```

```python
Dept2 = sns.displot(data=feedback,x='How satisfied were you with your Screening call? - de09e7f8-4e8c-498d-9ccb-237788b735d0',hue='Department',multiple='dodge')
plt.xticks(rotation=90)
```
