```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
```

```python
money = pd.read_csv('banknote-dimensions.csv')
```

```python
money.info()
```

```python
sns.lmplot(data=money,x='length',y='width',hue='currency')
```

```python
sns.lmplot(data=money,x='width',y='length',hue='code')
```

```python
sns.regplot(data=money,x='value',y='width')
```
