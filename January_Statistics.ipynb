```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
```

```python
JanFeedback = pd.read_csv('Jan Feedback.csv')
```

```python
JanFeedback.dtypes
```

```python
JanFeedback.info()
```

```python
statistic1 = sns.displot(data=JanFeedback,x='How satisfied were you with the ease of the application process? - 513b2b55-b818-4f03-bf6f-e82b180c9e60',hue='Location',multiple='stack')
statistic1.set(title='How satisfied were you with the ease of the application process?',ylabel='Frequency',xlabel='Scale')
plt.xticks(rotation=90)
```

```python
statistic2 = sns.displot(data=JanFeedback,x='How satisfied were you with the interview materials you received? - 4158aafc-ce92-4d35-930f-2f5451455a4d',hue='Location',multiple='stack')
statistic2.set(title='How satisfied were you with the interview materials you received? ',ylabel='Frequency',xlabel='Scale')
plt.xticks(rotation=90)
```

```python
satTwo = sns.displot(data=JanFeedback,x='How satisfied were you with the interview feedback you received? - e30211c6-d32d-4f89-95fc-dfb91bb47033',hue='Department',multiple='stack')
satTwo.set(title='How satisfied were you with the interview feedback you received? ',ylabel='Frequency',xlabel='Scale')
plt.xticks(rotation=90)
```

```python
satThree= sns.histplot(data=JanFeedback,x='How satisfied are you with the way you were treated? - 05c18285-2750-479f-8afc-d560b66f552c',hue='Department',multiple='stack')
satThree.set(title='How satisfied are you with the way you were treated?',ylabel='Frequency',xlabel='Scale')
plt.xticks(rotation=90)
```

```python
satFour = JanFeedback['Overall, how satisfied were you with your interviewing experience? - 693b0a81-d4e7-40a2-9135-7fc38fa43eaf'].value_counts()
```

```python
satFour.plot(kind='pie',legend=False,autopct='%1.1f%%',title='Overall, how satisfied were you with your interviewing experience?',ylabel='Frequency')
```
