```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
df = pd.read_excel('NCCG 2324 accounts workings data (1).xlsx',sheet_name='Check of bank statement')
```

```python
df.set_index('Period',inplace=True)
```

```python
df['Cash opening'].plot(kind='barh',xlabel='Amount (£)',color='orange',title='NCCG 23-24 Cash Opening')
```

```python
df['Receipts'].plot(kind='barh',color='violet',title='NCGG 23-24 Reciepts',xlabel='Amount (£)')
```

```python
df['Payments'].plot(kind='barh',color='blue',title='NCCG 23-24 Payments',xlabel='Amount (£)')
```

```python
df['Bank Cash closing'].plot(kind='barh',color='silver',title='NCCG 23-24 Cash Closing',xlabel='Amount (£)')
```
